Movie Duration

```python
movies=df[df["type"]=="Movie"].copy()
print(movies)
```

```python
movies['duration_min']=(movies['duration'].str.extract(r"(\d+)").astype(float))
print(movies['duration_min'])
```

```python
movies[["title","duration","duration_min"]].head()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(
    movies["duration_min"].dropna(),
    bins=35,
    kde=True
)
plt.title("Distribution of Movie Duration")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")
plt.show()
```

Movie Duration Boxplot

```python
plt.figure(figsize=(8,5))
plt.boxplot(
    x=movies["duration_min"]
)
plt.title("Movie Duration Distribution")
plt.xlabel("Duration (minutes)")
plt.show()
```

Content Added Over Time

```python
df['year_added']=df['date_added'].dt.year
df.isnull().sum()
df['year_added']=pd.to_datetime(df["year_added"],errors='coerce')
df.isnull().sum()
added_year=df['year_added'].value_counts().sort_index()
print(added_year)
```

```python
plt.figure(figsize=(10,6))
plt.plot(
    added_year.index,
    added_year.values,
    marker="o"
)
plt.title("Titles Added to Amazon Prime by Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")
plt.show()
```

Content Added by Month

```python
df['month_add']=df['date_added'].dt.month_name()
month_count=df['month_add'].value_counts()
plt.figure(figsize=(12,5))
sns.barplot(
    x=month_count.index,
    y=month_count.values
)
plt.title("Titles Added by Month")
plt.xlabel("Month")
plt.ylabel("Number of Titles")
#plt.xticks(rotation=45)
plt.show()
```
